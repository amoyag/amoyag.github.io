## 1. Objetivo y especificación · qué se modifica y qué se protege

**Escribid:** dos o tres frases con el objetivo y su razón; después completad la tabla de restricciones de la plantilla a partir de vuestra ficha.

**Preguntas para decidir:**

- ¿La lista tiene huecos? Conservad las posiciones reales, no el intervalo entre primera y última.

## 2. Evidencia y estrategia · por qué tomar esa decisión

**Escribid:** un párrafo que conecte la evidencia con la decisión, y otro breve sobre una estrategia alternativa que habéis considerado.

**Recuperad:** una estructura o figura pertinente y una fuente primaria identificada del dosier o materiales ya trabajados. Citad la referencia donde sostiene la afirmación. Si conocéis una afirmación solo por una fuente intermediaria, atribuidla así; no la presentéis como lectura propia del artículo original.

**Completad mentalmente:** «En [figura/fuente] observamos [dato]. Esto apoya [interpretación], por lo que elegimos [decisión]. No demuestra [límite]. Consideramos [alternativa], que descartamos por [razón ligada al objetivo]».

**Ejemplo de razonamiento, ajeno al proyecto:** una forma estática compatible con un acceso abierto permite plantear una hipótesis sobre acceso; por sí sola no demuestra movimiento ni actividad. La decisión necesita distinguir observación estructural y consecuencia funcional.

No llenéis el espacio con definiciones de AlphaFold, RFdiffusion o ProteinMPNN: explicad por qué vuestra estrategia responde a vuestro objetivo.

## 3. Éxito en tres niveles · qué tendría que ocurrir

**Escribid tres párrafos breves, con los rótulos de la plantilla:**

1. **Aceptación computacional:** qué medidas y condiciones permitirán aceptar provisionalmente un candidato. Fijadlas antes de examinar el pool propio y remitiros al panel del apartado 5.
2. **Hipótesis funcional y alternativa:** si la geometría predicha fuera correcta, qué comportamiento esperaríais y qué otra explicación podría confundirse con él.
3. **Experimento discriminante:** qué medir, contra qué control y qué resultados separarían ambas explicaciones. Se propone la prueba; no se pide ejecutarla.

**Comprobación:** si vuestro texto dice «el RMSD demuestra actividad», mezcláis niveles. Si dice solo «mediríamos actividad», faltan comparación, control e interpretación. Podéis referiros a ensayos estudiados, sin inventar protocolos detallados ni resultados.

## 4. Fallo anticipado · qué podría escapar al criterio

**Escribid un único fallo concreto y desarrolladlo.** Debe vincularse a vuestro objetivo, no ser una lista genérica de riesgos.

- ¿Qué región o interacción podría alterarse mientras una medida general sigue siendo aceptable?
- ¿Por qué el panel previsto podría no detectarlo?
- ¿Qué inspección o medida pertinente lo detectaría y sobre qué residuos?
- ¿Qué condición justificaría rechazar o revisar el candidato? Indicad procedencia del umbral o que se trata de una propuesta razonada.

**Ejemplo sin cifras, ajeno al proyecto:** una media estructural global puede ocultar un cambio en una región de acceso. Para examinar su posición, el ajuste debe apoyarse en una región de referencia declarada y la evaluación en la región de interés. Ajustar esa región sobre sí misma responde a otra pregunta.

«Puede perder función» no basta: explicad el fallo y qué evidencia lo mostraría.
